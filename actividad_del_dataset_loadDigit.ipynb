```python
import matplotlib.pyplot as plt
from sklearn import datasets, metrics
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB, BernoulliNB, MultinomialNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# Cargar dataset
digits = datasets.load_digits()

# Preprocesamiento (Aplanar las imágenes)
n_samples = len(digits.images)
data = digits.images.reshape((n_samples, -1))

# Dividir en entrenamiento y prueba (siguiendo la lógica de tus ejemplos)
X_train, X_test, y_train, y_test = train_test_split(
    data, digits.target, test_size=0.5, shuffle=False
)

# Diccionario con todos los modelos solicitados
algoritmos = {
    "Regresión Logística": LogisticRegression(max_iter=10000),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "SVM": SVC(gamma=0.001),
    "Naive Bayes Gaussiano": GaussianNB(),
    "Naive Bayes Bernoulli": BernoulliNB(),
    "Naive Bayes Multinomial": MultinomialNB(),
    "Árbol de Decisión": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(n_estimators=100)
}

# Ejecución y generación de reportes
for nombre, modelo in algoritmos.items():
    modelo.fit(X_train, y_train)
    predicciones = modelo.predict(X_test)

    print(f"\n--- {nombre.upper()} ---")
    print(metrics.classification_report(y_test, predicciones))

    # Matriz de Confusión
    disp = metrics.ConfusionMatrixDisplay.from_predictions(y_test, predicciones)
    disp.figure_.suptitle(f"Matriz de Confusión: {nombre}")
    plt.show()

# Visualización de los 5 dígitos (Captura solicitada)
_, axes = plt.subplots(nrows=1, ncols=5, figsize=(10, 3))
for ax, imagen, pred, real in zip(axes, X_test, predicciones, y_test):
    ax.set_axis_off()
    ax.imshow(imagen.reshape(8, 8), cmap=plt.cm.gray_r)
    ax.set_title(f"Real: {real}\nPred: {pred}")
```
